```python
import os
import sys
import pandas as pd

# ---- Settings ----
PROJECT_ROOT = r"D:\RUSL\Final Project\TTS\voicelk_ml"
INPUT_PATH = r"D:\RUSL\Final Project\TTS\voicelk_ml\data\prepared\speaker_01_cleaned.csv"
OUTPUT_PATH = r"D:\RUSL\Final Project\TTS\voicelk_ml\data\prepared\speaker_01_metadata.csv"

FILE_NAME_COL = 0   # 1st column: file name
SPEAKER_COL = 1     # 2nd column: speaker id
TEXT_COL = 2        # 3rd column: text

sys.path.insert(0, os.path.join(PROJECT_ROOT, "model_engine"))
from pipeline import TextProcessingPipeline

# ---- Load (no header row) ----
if INPUT_PATH.lower().endswith((".xlsx", ".xls")):
    df = pd.read_excel(INPUT_PATH, header=None, dtype=str)
else:
    df = pd.read_csv(INPUT_PATH, header=None, dtype=str, encoding="utf-8")

print(f"Rows: {len(df)}")

pipeline = TextProcessingPipeline()

# ---- Convert row by row ----
records, errors = [], []

for idx, row in df.iterrows():
    file_name = str(row.iloc[FILE_NAME_COL]).strip()
    speaker_id = str(row.iloc[SPEAKER_COL]).strip()
    text = row.iloc[TEXT_COL]

    if pd.isna(text) or not str(text).strip():
        records.append([file_name, speaker_id, "", ""])
        continue

    try:
        result = pipeline.process(str(text).strip())
        records.append([file_name, speaker_id, result["normalized_text"], result["ipa_sequence"]])
    except Exception as e:
        records.append([file_name, speaker_id, "", ""])
        errors.append((idx, str(e)))

    if (idx + 1) % 100 == 0:
        print(f"Processed {idx + 1} rows...")

# ---- Output DataFrame: file_name, speaker_id, normalized_text, ipa_sequence ----
out_df = pd.DataFrame(records, columns=["file_name", "speaker_id", "normalized_text", "ipa_sequence"])

print(f"\nDone. Errors: {len(errors)}")
for idx, msg in errors[:20]:
    print(f"  row {idx}: {msg}")

# Save without header and without index
out_df.to_csv(OUTPUT_PATH, index=False, header=False, encoding="utf-8")
print(f"Saved to {OUTPUT_PATH}")

out_df.head()
```
